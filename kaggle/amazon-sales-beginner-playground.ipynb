Exercise 1: Clean & Convert rating and rating_count
Right now, rating and rating_count are stored as text (objects) because of non-numeric characters and commas.

> Task:

> Convert rating to float (Watch out: there might be invalid text entries like '|'—use pd.to_numeric(..., errors='coerce')).

> Clean commas from rating_count, fill missing values (fillna(0)), and convert to int.

```python
# Convert rating to numeric (setting invalid values to NaN)

#pd.numeric - converts values into float
#errors - argument is used to bypass any errors in reading data
#eg.eny strings instead to numbers ('one' instead of '1' will throw error)
df['rating_clean'] = pd.to_numeric(df['rating'], errors='coerce')
```

```python
# Clean rating_count

#.str.replace(',', ''): Removes commas(ex:"24,269" -> "24269").
#.fillna('0'): Replaces any missing/null (NaN) values with '0'
#.astype(int): Converts the cleaned string values into integer
df['rating_count_clean'] = (
    df['rating_count']
    .str.replace(',', '')
    .fillna('0')
    .astype(int)
)
```

```python
df[['rating_clean', 'rating_count_clean']].info()
```

```python
# DEMO: Handling Missing Values (NaN) with fillna(), ffill(), bfill(), & interpolate()

# 1. Prepare a small sample subset with missing values to visualize clearly
missing_demo_df = pd.DataFrame({
    'product_id': df['product_id'].head(7),
    'original_rating': df['rating'].head(7)
})

# Coerce non-numeric values to NaN so we have missing values to test
missing_demo_df['rating_raw'] = pd.to_numeric(missing_demo_df['original_rating'], errors='coerce')

# Intentionally insert a couple of extra NaNs for demonstration purposes
missing_demo_df.loc[2:3, 'rating_raw'] = np.nan

print("--- Original Sample Data with Missing Values ---")
print(missing_demo_df[['product_id', 'rating_raw']])
print("\n" + "="*50 + "\n")


# 2. Strategy A: Static Value Fill using .fillna()
# Best for categorical data or when 0 / fixed defaults make logical sense
missing_demo_df['filled_static'] = missing_demo_df['rating_raw'].fillna(0.0)


# 3. Strategy B: Directional Fills (.ffill() and .bfill())
# Best when consecutive data points carry over from adjacent rows
missing_demo_df['forward_filled'] = missing_demo_df['rating_raw'].ffill()  # Propagates last valid observation forward
missing_demo_df['backward_filled'] = missing_demo_df['rating_raw'].bfill() # Propagates next valid observation backward


# 4. Strategy C: Linear Interpolation (.interpolate())
# Best for continuous numeric sequences (calculates mathematical midpoint between missing gaps)
missing_demo_df['interpolated_linear'] = missing_demo_df['rating_raw'].interpolate(method='linear')


# 5. Display comparative results side-by-side
print("--- Comparative Summary of Imputation Methods ---")
print(missing_demo_df[[
    'rating_raw', 
    'filled_static', 
    'forward_filled', 
    'backward_filled', 
    'interpolated_linear'
]])
```

```python
# ==============================================================================
# DEMO: Exploring Different Interpolation Methods in pandas
# ==============================================================================

interp_df = missing_demo_df[['product_id', 'rating_raw']].copy()

# 1. Linear: Equal steps along a straight line between known neighbors
interp_df['linear'] = interp_df['rating_raw'].interpolate(method='linear')

# 2. Nearest: Copies the closest valid neighbor (uses index distance for ties)
interp_df['nearest'] = interp_df['rating_raw'].interpolate(method='nearest')

# 3. Polynomial (order=2): Fits a quadratic curve through surrounding points
interp_df['polynomial_2'] = interp_df['rating_raw'].interpolate(method='polynomial', order=2)

# 4. Spline (order=2): Fits a piecewise smooth spline curve
interp_df['spline_2'] = interp_df['rating_raw'].interpolate(method='spline', order=2)

# 5. PCHIP: Shape-preserving piecewise cubic (avoids overshooting/wiggles)
interp_df['pchip'] = interp_df['rating_raw'].interpolate(method='pchip')

# 6. Limit direction: Control whether gaps at the very start/end get filled
edge_gap_series = pd.Series([np.nan, 4.0, np.nan, 4.4, np.nan])
edge_compare = pd.DataFrame({
    'raw': edge_gap_series,
    'both': edge_gap_series.interpolate(limit_direction='both'),
    'forward_only': edge_gap_series.interpolate(limit_direction='forward'),
    'backward_only': edge_gap_series.interpolate(limit_direction='backward'),
})

print("--- Interpolation Methods Comparison ---")
print(interp_df.round(3))

print("\n--- Edge NaNs with limit_direction ---")
print(edge_compare.round(3))
```

```python
# Exercise 2: Category Aggregations & Grouping

#Explore how product categories perform using `.groupby()`.

#Task:
#1. Find the average discount percentage per top-level category.
#2. Count how many products are in each category.
#3. Find the category with the **highest average product rating.


# Extract top-level category (before the first '|' split)
df['main_category'] = df['category'].str.split('|').str[0]

# Group by main category and get statistics

#.groupby('main_category'): 
#Buckets all 1465 rows of df into groups based on main_category

#.agg(...): Runs 'count' or 'mean' 
#separately on each group for each specified column,
#assigning the custom output column names you chose

###
#What is the role of .reset_index()?
#By default, .groupby('main_category') turns 'main_category'
#into the DataFrame index, rather than keeping it as a regular column.
###

df.groupby('main_category').agg(
    total_products=('product_id', 'count'),
    avg_actual_price=('actual_price_clean', 'mean'),
    avg_discounted_price=('discounted_price_clean', 'mean'),
    avg_rating=('rating_clean', 'mean')
).reset_index()
```

```python
### Exercise 3: Filter & Multi-Condition Querying

# /*Practice filtering DataFrames using conditional expressions or `.query()`.

# * Task: Find all products that:
# * Have a `rating_clean` greater than or equal to 4.5
# * Have at least 10,000 ratings (`rating_count_clean` greater than or equal to 10000)
# * Offer a discount greater than 50%

# Clean discount percentage to float first
df['discount_pct_clean'] = df['discount_percentage'].str.replace('%', '').astype(float)

# Filter using boolean indexing
popular_high_rated = df[
    (df['rating_clean'] >= 4.5) & 
    (df['rating_count_clean'] >= 10000) & 
    (df['discount_pct_clean'] > 50)
]

# View key columns
popular_high_rated[['product_name', 'actual_price_clean', 'discounted_price_clean', 'rating_clean']]
```

```python
### Exercise 4: Text Manipulation & Length Analysis

# Use Pandas string methods (`.str`) to analyze text patterns.

# * **Task:**
# 1. Find the top 5 most frequent words in `review_title`.
# 2. Check if longer product titles (`product_name`) correlate with higher ratings.


# Calculate character length of product names
df['name_length'] = df['product_name'].str.len()

# Compare average rating by title length quartile

# ***** pd.qcut(df['name_length'], q=4) ****
# qcut stands for quantile-based discretization.
# It splits your numeric column name_length into 4 equal-sized bins (quartiles)
# so that roughly 25% of the products fall into each bin.
df_quantiles = df.groupby(pd.qcut(df['name_length'], q=4),observed=False)['rating_clean'].mean()

print(df_quantiles)
```

```python
#df_quantiles with displaying number of products in each buckets using agg()
# Group by title length quartiles and inspect both product count and average rating
title_length_summary = (
    df.groupby(pd.qcut(df['name_length'], q=4), observed=False)
    .agg(
        product_count=('rating_clean', 'count'),
        avg_rating=('rating_clean', 'mean')
    )
    .reset_index()
)

title_length_summary
```

```python
# same df_quantiles with displaying number of products in each buckets using cut() instead of qcut() and agg()
# Group by title length intervals using pd.cut()
#pdcut() uses bins instead of q in argument

title_length_cut_summary = (
    df.groupby(pd.cut(df['name_length'], bins=4), observed=False)
    .agg(
        product_count=('rating_clean', 'count'),
        avg_rating=('rating_clean', 'mean')
    )
    .reset_index()
)
title_length_cut_summary
```

```python
### Exercise 5: Calculate Total Savings per Product

# Add new calculated columns to analyze price differences.

# * **Task:**
# 1. Compute the absolute discount amount in Rupees (`actual_price_clean - discounted_price_clean`).
# 2. Find the top 5 products with the **biggest absolute rupee savings**.


df['savings_in_rs'] = df['actual_price_clean'] - df['discounted_price_clean']

# Top 5 products with maximum Rupee discount

df.nlargest(5, 'savings_in_rs')

# Scans the savings_in_rs column, sorts rows internally in descending order, 
# and returns the top 5 rows with the highest discount amounts.
df.nlargest(5, 'savings_in_rs')[['product_name', 'actual_price_clean', 'discounted_price_clean', 'savings_in_rs']]
```
