# **Problem Solving Tasks**

```python
#get Input Numbers function commonly used for
#1)LArgest number program and
#2)frequency count program and
#3)Use a set to remove duplicate elements from a list.
def getInputNumbers():
    size = int( input('Enter the number of element'))
    iteration=0
    list_input=[]
    for i in range(0,size,1):
      list_input.append(int(input('Enter element: ')))
    return list_input
```

# *1)Find the Largest NUmber in the list*

```python
def findLargestNumberInList():
  numbers = getInputNumbers()
  if(len(numbers)==0):
    print('There are no values added')
  elif(len(numbers)==1):
    print('The only and maximum number is: ',numbers[0])
  else:
    max=numbers[0]
    for num in numbers:
      if(num>max):
        max=num
    print('The Largest number is: ',max)
```

```python
findLargestNumberInList()
```

# *Program to count the frequency of elements in a list.*

```python
def countFrequency():
  numbers = getInputNumbers()
  dict_numbers = dict()
  for num in numbers:
    if num not in dict_numbers:
      dict_numbers[num]=1
    else:
      dict_numbers[num] = dict_numbers[num]+1
  print('Number','->','Count','\n-------------')
  for key in dict_numbers:
    print(key,'->',dict_numbers[key],'times')
```

```python
countFrequency()
```

# ***3)Create a dictionary that stores student names as keys and marks as values, then find the student with the highest marks.*** #

```python
def getStudentDetails():
  studentDict = dict()
  noOfStudents = int(input('Enter Number of students (add atleast1 student)'))
  for i in range(0,noOfStudents):
    print('\nStudent No: ',(i+1))
    studentName = input('Enter Student Name:')
    mark = int(input('Enter Mark secured by the student:'))
    studentDict[studentName] = mark
  return studentDict
```

```python
import math
def findStudentWithMaxmark():
  studentDetails = getStudentDetails()
  studentName = ''
  maxMark = -math.inf
  for key in studentDetails.keys():
    if(studentDetails[key]>maxMark):
      studentName = key
      maxMark = studentDetails[key]
  print('The student with Maximum mark is:',studentName,' Mark:',studentDetails[studentName])
```

```python
#Assuming 1)each student has unique mark
#         2)Marks entered are 0 to 100
findStudentWithMaxmark()
```

# *Fibonacci Series*

```python
def fibonacci():
  number = int(input('Enter number of Fibonacci numbers needed start from 1: '))
  fibonacciSeries = [0]
  if(number==2):
    fibonacciSeries.append(1)
  a=0
  b=1
  iteration=2
  while(iteration<=number):
    c=a+b
    fibonacciSeries.append(c)
    a=b
    b=c
    iteration+=1
  print(fibonacciSeries)
```

```python
fibonacci()
```

# *Use a set to remove duplicate elements from a list.*

```python
def removeDuplicates():
  numbers = [] #initializing numbers list
  numbers = getInputNumbers()
  print('Initial list: ',numbers)

  #numbers contains duplicate values
  #we can type cast into set to get unique values and removes duplicate
  uniqueValues = set(numbers)
  print(uniqueValues)

  #assigning that set back to list using list() datatype casting
  numbers = list(uniqueValues)
  print('FInal List',numbers)
  print(type(numbers))
```

```python
removeDuplicates()
```
