```python
tuple1 = ('apple','banana','cherry')
tuple2 = (1,2,3,4)
tuple3 = ('text',42,True)
```

```python
print(tuple1,tuple2,tuple3)
```

```python
#Accessing the element in the tuple
#accessing apple
tuple1[0]
```

```python
#access cherry
tuple1[2]
```

```python
tuple1[-1]
```

```python
#access 'text' and 42
tuple3[0:2]
```

```python
#Access 2,3,4
tuple2[1:]
```

```python
#type
type(tuple2)
```
