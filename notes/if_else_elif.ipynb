# **If Statement**

```python
age = 20
age >=18
```

```python
if age >= 18:
  print('You are eligible to vote')
```

# **If else statement**

```python
age=10
if age >= 18:
  print('You are eligible to vote')
else:
  print('You are eligible to vote')
```

```python
#grading system
# >=90 -> A ,75 - 90 'B', 60 to 75 'C' ,35 - 60 -> 'D'
# 20 to 35 'E'
# 0 to 20 'F'
score = 85
if score >= 90:
  print('A')
elif score >= 75:
  print('B')
elif score >= 60:
  print('C')
elif score >= 35:
  print('D')
elif score >= 20:
  print('E')
else:
  print('F')
```

```python
#nested condition
age = 25
indian_citizen = True
if age>=18:
  if indian_citizen:
    print('Person is indian citizen and >=18 years, So eligible to vote')
  else:
    print('Person > 18 but not an indian citizen')
else:
  print('Not eligible to vote')
```
